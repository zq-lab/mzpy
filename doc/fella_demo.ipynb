## 富集分析

```python
from mzpy.fella import Fella

fella = Fella(organism="dre", h5_path="./data/dre.h5",
              load_matrix=["diffusion", "pagerank"])

fella.enrich(compounds=["C00002", "C00040", "C00186"], methods="diffusion")

# 第一次计算并缓存
df = fella.generate_results_table()

# 第二次直接返回缓存，不重复计算
df2 = fella.generate_results_table()

# 换参数则自动重新计算
df3 = fella.generate_results_table(threshold=0.01)

# 绘图也会复用已缓存的 graph
fella.plot(target="result.png")

# 酶基因列表
df_enzymes = fella.generate_enzymes_table(threshold=0.05)

# 查看当前状态
print(fella.summary())
```
